```python
# чистка данных
import pandas as pd
import pathlib, sys
ROOT = pathlib.Path(r'C:\Users\artur\Desktop\питон\1')
sys.path.insert(0, str(ROOT))
from src.cleaning import *
from src.config import DATA_INTERIM, DATA_PROCESSED
```

```python
# читаю склеенный файл
df = pd.read_parquet(DATA_INTERIM / 'source.parquet')
print('на входе:', df.shape)
print(df.head(2))
```

```python
# переименовываю столбцы и убираю пробелы
df = rename_columns(df)
df = strip_strings(df)
df = normalize_categories(df)
print('столбцы теперь:', list(df.columns))
```

```python
# чиню значения (даты экселя) и перевожу в числа
df = clean_text(df)
df = coerce_value(df)
print('пустых значений:', int(df['value'].isna().sum()))
print(df['value'].describe())
```

```python
# год в число
df = coerce_year(df)
print(df['year'].describe())
```

```python
# удаляю дубликаты
df = drop_full_duplicates(df)
df = drop_key_duplicates(df)
print('осталось строк:', len(df))
```

```python
# помечаю выбросы флажком (не удаляю!)
df = mark_outliers(df)
print(df['is_outlier'].value_counts())
```

```python
# ставлю экономные типы, удаляю сноску
df = apply_schema(df)
print(df.dtypes)
print('память МБ:', round(df.memory_usage(deep=True).sum() / 1024 / 1024, 1))
df.to_parquet(DATA_PROCESSED / 'clean.parquet', index=False)
print('сохранил, строк:', len(df))
```
