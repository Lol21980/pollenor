```python
import pandas as pd
df = pd.read_csv('sample_200k.csv', sep=';', encoding='utf-8-sig', engine='python')
print(df.shape)
print(df.dtypes)
print((df.isna().mean()*100).round(2))
```

```python
import matplotlib.pyplot as plt
df['Год'].value_counts().sort_index().plot()
plt.title('Записей по годам (срез 200к)')
plt.xlabel('год')
plt.show()
```
